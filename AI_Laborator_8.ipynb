## 📚  Aspecte teoretice

Chiar daca procesul prelucrarii datelor este similar celui din cazul imaginilor, in cazul textelor se adauga o serie de pasi de pre-procesare, adesea vitali, si de reprezentare a textelor.

Metode de pre-procesare:
- stemming & lemmatisation - aducerea cuvintelor la forma canonica
- curatarea textului de caractere speciale si semne de punctuatie
- eliminarea cifrelor/numerelor din cadrul textului
- corectarea greselilor de scriere (ex. "teh" -> "the")
- expansiunea contractiilor (ex. "it's" -> "it is", "can't" -> "can not")

Metode de reprezentare:
- Bag of Words (BoW) - fiecare cuvant este reprezentat printr-un vector binar de dimensiune egala cu numarul de cuvinte din vocabularul corpus-ului
- Term Frequency-Inverse Document Frequency (TF-IDF) - fiecare cuvant este reprezentat printr-un vector de dimensiune egala cu numarul de cuvinte din vocabularul corpus-ului, dar fiecare cuvant este ponderat in functie de frecventa sa in corpus
- Word2Vec features (embeddings) - fiecare cuvant este reprezentat printr-un vector (cu valori reale) de dimensiune fixa (ex. 100, 200, 300) care contine informatii despre semantica sa

```python
import gensim
import os
import numpy as np

# load Google pretrained word2vec model (you must have this file locally)
model_path = '/content/drive/MyDrive/ColabData/GoogleNews-vectors-negative300.bin'

word2vecModel300 = gensim.models.KeyedVectors.load_word2vec_format(model_path, binary=True)

# check some known words
print(word2vecModel300.most_similar('support'))
print("vector for 'room':", word2vecModel300["room"])

# check if a word exists in the model
word = "cuvantciudat"
if word in word2vecModel300.index_to_key:
    print("vector for 'cuvantciudat':", word2vecModel300[word])
else:
    print("word cuvantciudat not found in vocabulary")

# function to compute vector features for full sentences
def featureComputation(model, data):
    features = []
    # split each sentence into words
    phrases = [phrase.split() for phrase in data]
    for phrase in phrases:
        # keep only known words longer than 2 characters
        vectors = [model[word] for word in phrase if len(word) > 2 and word in model.index_to_key]
        # if no known words, return zero vector
        if len(vectors) == 0:
            result = [0.0] * model.vector_size
        else:
            # average the vectors of the known words
            result = np.mean(vectors, axis=0)
        features.append(result)
    return features

# compute features using the pretrained model
trainFeatures = featureComputation(word2vecModel300, trainInputs)
testFeatures = featureComputation(word2vecModel300, testInputs)
```

```python
trainFeatures_word2vec = featureComputation(word2vecModel300, trainInputs)
testFeatures_word2vec = featureComputation(word2vecModel300, testInputs)
```

```python
# convert features into numpy arrays so they can be used by the ANN
X_train = np.array(trainFeatures_word2vec)
X_test = np.array(testFeatures_word2vec)

# convert output labels into numerical format ('positive' -> 1, 'negative' -> 0)
y_train = np.array([labelNames.index(label) for label in trainOutputs])
y_test = np.array([labelNames.index(label) for label in testOutputs])
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
```

```python
# Build the ANN model

# create a sequential model (layer by layer ANN)
model = Sequential()
# add the first hidden layer with 64 neurons and ReLU activation
# input_dim is the number of features for each sample
model.add(Dense(64, input_dim=X_train.shape[1], activation='relu'))
# add a second hidden layer with 32 neurons and ReLU activation
model.add(Dense(32, activation='relu'))
# add the output layer with as many neurons as label classes: 2
# use softmax to get probabilities for each class
model.add(Dense(len(labelNames), activation='softmax'))
```

```python
from tensorflow.keras.optimizers import Adam

# define the optimizer and set the learning rate
optimizer = Adam(learning_rate=0.0005)

# compile the model: use sparse categorical crossentropy because labels are integers
# track accuracy as the evaluation metric
model.compile(loss='sparse_categorical_crossentropy', optimizer=optimizer, metrics=['accuracy'])


# train the model with training data, 20% for validation
model.fit(X_train, y_train, epochs=20, batch_size=64, validation_split=0.2)
```

```python
# use the trained model to predict labels for test data
predictions = model.predict(X_test)
# convert the predicted probabilities into class indices
predicted_classes = np.argmax(predictions, axis=1)

# print predictions and compare with actual labels
for i in range(len(testInputs)):
    if predicted_classes[i] == y_test[i]:
        print(f"Input: {testInputs[i]} -> Predicted: {labelNames[predicted_classes[i]]} -> Correct!")
    else:
        print(f"Input: {testInputs[i]} -> Predicted: {labelNames[predicted_classes[i]]} -> Incorrect! (True label: {labelNames[y_test[i]]})")

# accuracy score
print("Accuracy: ", accuracy_score(y_test, predicted_classes))
```

## Antrenarea unui clasificator bazat pe ANN (cod propriu) si stabilirea sentimentului mesajului

```python
import numpy as np

class ANN:
    def __init__(self, input_size, hidden_size, output_size, learning_rate=0.01):
        self.input_size = input_size
        self.hidden_size = hidden_size
        self.output_size = output_size
        self.learning_rate = learning_rate

        # initialize weights and biases for input-hidden and hidden-output layers
        self.W1 = np.random.randn(input_size, hidden_size).astype(np.float32)
        self.b1 = np.zeros((1, hidden_size)).astype(np.float32)
        self.W2 = np.random.randn(hidden_size, output_size).astype(np.float32)
        self.b2 = np.zeros((1, output_size)).astype(np.float32)

    # activation functions
    def relu(self, x):
        return np.maximum(0, x)

    def relu_derivative(self, x):
        return (x > 0).astype(float)

    def softmax(self, x):
        # prevent overflow using max subtraction (numerical stability)
        exp_x = np.exp(x - np.max(x, axis=1, keepdims=True))
        return exp_x / np.sum(exp_x, axis=1, keepdims=True)

    # loss function: categorical cross-entropy
    def categorical_crossentropy(self, y_true, y_pred):
        m = y_true.shape[0]
        return -np.sum(y_true * np.log(y_pred + 1e-10)) / m  # added epsilon to prevent log(0)

    # derivative of the loss function
    def categorical_crossentropy_derivative(self, y_true, y_pred):
        return y_pred - y_true

    # forward pass through the network
    def forward(self, X):
        Z1 = np.dot(X, self.W1) + self.b1
        A1 = self.relu(Z1)
        Z2 = np.dot(A1, self.W2) + self.b2
        A2 = self.softmax(Z2)
        return A1, A2

    # backward pass and parameter updates
    def backprop(self, X, y, A1, A2):
        m = X.shape[0]

        # compute gradient of loss with respect to output layer activations
        dA2 = self.categorical_crossentropy_derivative(y, A2)

        # gradients for weights and biases in the output layer
        dW2 = np.dot(A1.T, dA2) / m
        db2 = np.sum(dA2, axis=0, keepdims=True) / m

        # backpropagate into the hidden layer
        dA1 = np.dot(dA2, self.W2.T)
        dZ1 = dA1 * self.relu_derivative(A1)

        # gradients for weights and biases in the hidden layer
        dW1 = np.dot(X.T, dZ1) / m
        db1 = np.sum(dZ1, axis=0, keepdims=True) / m

        # update weights and biases
        self.W1 -= self.learning_rate * dW1
        self.b1 -= self.learning_rate * db1
        self.W2 -= self.learning_rate * dW2
        self.b2 -= self.learning_rate * db2

    # training function
    def train(self, X_train, y_train, X_test, y_test, epochs=100):
        for epoch in range(epochs):
            # forward pass
            A1, A2 = self.forward(X_train)

            # calculate training loss
            loss = self.categorical_crossentropy(y_train, A2)

            # backward pass and update weights
            self.backprop(X_train, y_train, A1, A2)

            # print loss and accuracy every 10 epochs
            if epoch % 10 == 0:
                train_preds = self.predict(X_train)
                train_acc = self.accuracy(np.argmax(y_train, axis=1), train_preds)

                test_preds = self.predict(X_test)
                test_acc = self.accuracy(np.argmax(y_test, axis=1), test_preds)

                print(f'Epoch {epoch}, Loss: {loss:.4f}, Train Accuracy: {train_acc:.4f}, Test Accuracy: {test_acc:.4f}')

        # final predictions on the test set
        predictions = self.predict(X_test)
        return predictions

    # prediction function for new data
    def predict(self, X):
        _, A2 = self.forward(X)
        return np.argmax(A2, axis=1)

    # accuracy calculation
    def accuracy(self, y_true, y_pred):
        return np.mean(y_true == y_pred)

# utility function for data preparation
def prepare_data(inputs, outputs, label_names):
    # ensure inputs is a 2D array
    X = np.array(inputs)
    if X.ndim == 1:
        X = X.reshape(-1, 1)

    # convert labels to numeric indexes
    y = np.array([label_names.index(label) for label in outputs])

    # one-hot encode labels
    y_one_hot = np.zeros((y.size, y.max() + 1))
    y_one_hot[np.arange(y.size), y] = 1

    return X.astype(np.float32), y_one_hot.astype(np.float32)

# check training and testing shapes
print(X_train.shape, y_train.shape)
print(X_test.shape, y_test.shape)

# prepare training and test data (replace with your actual feature variables)
X_train, y_train = prepare_data(trainFeatures_word2vec, trainOutputs, labelNames)
X_test, y_test = prepare_data(testFeatures_word2vec, testOutputs, labelNames)

# define ANN architecture
input_size = X_train.shape[1]
hidden_size = 64
output_size = len(labelNames)

# create and train the neural network
ann = ANN(input_size, hidden_size, output_size, learning_rate=0.01)
predictions = ann.train(X_train, y_train, X_test, y_test)
```

```python
# get predictions for testFeatures_word2vec
computed_test_indices = ann.predict(X_test)  # returns class indices
computed_test_outputs = [labelNames[idx] for idx in computed_test_indices]  # convert indices to labels

# display the results for each example
for i in range(len(testInputs)):  # testInputs = original texts
    predicted = computed_test_outputs[i]
    true_label = testOutputs[i]

    if predicted == true_label:
        print(f"input: {testInputs[i]} -> predicted: {predicted} -> correct!")
    else:
        print(f"input: {testInputs[i]} -> predicted: {predicted} -> incorrect! (true label: {true_label})")

# calculate accuracy on the test set
test_preds = computed_test_indices
y_test_labels = np.argmax(y_test, axis=1)
test_accuracy = ann.accuracy(y_test_labels, test_preds)
print("Test accuracy (final model):", test_accuracy)
```
